```python
a=[-2,-10,0,8,3]
mini=a[0]
maxi=a[0]
for i in range(len(a)):
  if(a[i]>maxi):
    maxi=a[i]
  elif(a[i]<mini):
    mini=a[i]
print("max:",maxi)
print("min:",mini)
```
